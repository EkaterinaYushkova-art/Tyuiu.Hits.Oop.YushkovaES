#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class ZooAnimal
{
    private string _name;
    private int _age;

    public string Name
    {
        get => _name;
        set => _name = value;
    }

    public string Species { get; set; }

    public int Age
    {
        get => _age;
        set
        {
            if (value < 0) throw new ArgumentOutOfRangeException(nameof(value));
            _age = value;
        }
    }

    public ZooAnimal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
    }

    public virtual void Feed(string food)
    {
        Console.WriteLine($"{Name} получил корм: {food}");
    }
}

public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, bool hasFur)
        : base(name, species, age)
    {
        HasFur = hasFur;
    }
}

public class Bird : ZooAnimal
{
    public bool CanFly { get; set; }

    public Bird(string name, string species, int age, bool canFly)
        : base(name, species, age)
    {
        CanFly = canFly;
    }
}

public class Reptile : ZooAnimal
{
    public double BodyTemp { get; set; }

    public Reptile(string name, string species, int age, double bodyTemp)
        : base(name, species, age)
    {
        BodyTemp = bodyTemp;
    }
}

public class Enclosure
{
    public string Title { get; set; }
    private readonly List<ZooAnimal> animals = new();

    public Enclosure(string title) => Title = title;

    public IReadOnlyList<ZooAnimal> Animals => animals;

    public void Place(ZooAnimal animal)
    {
        animals.Add(animal);
        Console.WriteLine($"{animal.Name} помещён в вольер «{Title}»");
    }
}

public class Zoo
{
    public string Name { get; set; }
    private readonly List<Enclosure> enclosures = new();

    public Zoo(string name) => Name = name;

    public void AddEnclosure(Enclosure enclosure) => enclosures.Add(enclosure);

    public void FeedAll(string food)
    {
        foreach (var enclosure in enclosures)
            foreach (var animal in enclosure.Animals)
                animal.Feed(food);
    }
}

var zoo = new Zoo("ZooCare");
var enclosure = new Enclosure("Тропики");
enclosure.Place(new Mammal("Балу", "Медведь", 7, true));
enclosure.Place(new Bird("Рио", "Попугай", 2, true));
enclosure.Place(new Reptile("Гена", "Крокодил", 15, 28.5));
zoo.AddEnclosure(enclosure);
zoo.FeedAll("фрукты и мясо");
```
