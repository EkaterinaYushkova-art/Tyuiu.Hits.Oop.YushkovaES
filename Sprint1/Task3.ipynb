### Система учёта животных в зоопарке (ZooCare)

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект **ZooCare** расширен: у `ZooAnimal`, `Mammal`, `Bird`, `Reptile`, `Enclosure` и `Zoo` добавлены конструкторы, свойства с геттерами/сеттерами и взаимодействие объектов (животное ↔ вольер ↔ зоопарк).

#### Дополнительное задание
Добавьте к существующим классам конструкторы классов с использованием геттеров и сеттеров и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class ZooAnimal
{
    private string _name;
    private int _age;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public string Species { get; set; }

    public int Age
    {
        get { return _age; }
        set { _age = value < 0 ? 0 : value; }
    }

    public ZooAnimal()
    {
        Name = "Безымянный";
        Species = "Неизвестно";
        Age = 0;
    }

    public ZooAnimal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт звук");
    }

    public virtual string GetInfo() => $"{Species} {Name}, {Age} лет";
}

public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, bool hasFur)
        : base(name, species, age)
    {
        HasFur = hasFur;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: звук млекопитающего");
}

public class Bird : ZooAnimal
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double wingspan)
        : base(name, species, age)
    {
        Wingspan = wingspan;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: чирик-чирик");
}

public class Reptile : ZooAnimal
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, bool isVenomous)
        : base(name, species, age)
    {
        IsVenomous = isVenomous;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: ш-ш-ш");
}

public class Enclosure
{
    public string Title { get; set; }
    public List<ZooAnimal> Animals { get; private set; } = new List<ZooAnimal>();

    public Enclosure(string title)
    {
        Title = title;
    }

    public void AddAnimal(ZooAnimal animal)
    {
        Animals.Add(animal);
        Console.WriteLine($"«{Title}»: принял {animal.GetInfo()}");
    }

    public void FeedAll()
    {
        Console.WriteLine($"Кормление в вольере «{Title}»:");
        foreach (var a in Animals)
        {
            Console.WriteLine($"  кормим {a.Name}");
            a.MakeSound();
        }
    }
}

public class Zoo
{
    public string Name { get; set; }
    public List<Enclosure> Enclosures { get; private set; } = new List<Enclosure>();

    public Zoo(string name)
    {
        Name = name;
    }

    public void AddEnclosure(Enclosure enclosure) => Enclosures.Add(enclosure);

    public void Transfer(ZooAnimal animal, Enclosure from, Enclosure to)
    {
        if (from.Animals.Remove(animal))
        {
            to.AddAnimal(animal);
            Console.WriteLine($"Перемещение {animal.Name}: {from.Title} → {to.Title}");
        }
    }

    public void Show()
    {
        Console.WriteLine($"=== {Name} ===");
        foreach (var e in Enclosures)
        {
            Console.WriteLine($"Вольер {e.Title}: {e.Animals.Count} животных");
            foreach (var a in e.Animals)
                Console.WriteLine("  " + a.GetInfo());
        }
    }
}

Zoo zoo = new Zoo("ZooCare Park");
Enclosure m = new Enclosure("Млекопитающие");
Enclosure b = new Enclosure("Птицы");
Enclosure r = new Enclosure("Рептилии");

var dog = new Mammal("Рекс", "Собака", 5, true);
var cat = new Mammal("Мурка", "Кошка", 3, true);
var parrot = new Bird("Кеша", "Попугай", 2, 0.4);
var snake = new Reptile("Каа", "Питон", 7, false);

m.AddAnimal(dog);
m.AddAnimal(cat);
b.AddAnimal(parrot);
r.AddAnimal(snake);

zoo.AddEnclosure(m);
zoo.AddEnclosure(b);
zoo.AddEnclosure(r);

m.FeedAll();
zoo.Transfer(cat, m, b);
zoo.Show();
```
