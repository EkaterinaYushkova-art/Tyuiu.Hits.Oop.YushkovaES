#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
public delegate void ZooEventHandler(string message);

public abstract class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }

    public event ZooEventHandler OnFed;
    public event ZooEventHandler OnAlert;

    protected ZooAnimal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
    }

    public virtual void Describe() => Console.WriteLine($"{Name} ({Species}), {Age} лет");

    public virtual void Feed(string food)
    {
        OnFed?.Invoke($"{Name} получил {food}");
    }

    public void RaiseAlert(string reason)
    {
        OnAlert?.Invoke($"{Name}: {reason}");
    }
}

public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, bool hasFur)
        : base(name, species, age) => HasFur = hasFur;
}

public class Bird : ZooAnimal
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double wingspan)
        : base(name, species, age) => Wingspan = wingspan;
}

public class Reptile : ZooAnimal
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, bool isVenomous)
        : base(name, species, age) => IsVenomous = isVenomous;
}

public class Zoo
{
    public string Name { get; set; }
    public List<ZooAnimal> Animals { get; } = new();
    public Dictionary<string, ZooAnimal> ByName { get; } = new();
    public HashSet<string> SpeciesRegistry { get; } = new();

    public event ZooEventHandler OnZooMessage;

    public Zoo(string name) => Name = name;

    public void Add(ZooAnimal animal)
    {
        Animals.Add(animal);
        ByName[animal.Name] = animal;
        SpeciesRegistry.Add(animal.Species);
        OnZooMessage?.Invoke($"Добавлен: {animal.Name}");
    }

    public void FeedAll(string food)
    {
        foreach (var animal in Animals)
            animal.Feed(food);
    }
}

var zoo = new Zoo("ZooCare");
zoo.OnZooMessage += msg => Console.WriteLine($"[Зоопарк] {msg}");

var lion = new Mammal("Симба", "Лев", 5, true);
var ostrich = new Bird("Киви", "Страус", 3, 2.1);
var croc = new Reptile("Гена", "Крокодил", 12, false);

lion.OnFed += msg => Console.WriteLine($"[Кормление] {msg}");
ostrich.OnFed += msg => Console.WriteLine($"[Кормление] {msg}");
croc.OnAlert += msg => Console.WriteLine($"[Тревога] {msg}");

zoo.Add(lion);
zoo.Add(ostrich);
zoo.Add(croc);

Console.WriteLine($"Всего животных: {zoo.Animals.Count}");
Console.WriteLine($"Уникальных видов: {zoo.SpeciesRegistry.Count}");
foreach (var species in zoo.SpeciesRegistry)
    Console.WriteLine($" - {species}");

if (zoo.ByName.TryGetValue("Симба", out var found))
    found.Describe();

zoo.FeedAll("рацион дня");
croc.RaiseAlert("нужен осмотр ветеринара");

// делегат-фильтр
ZooEventHandler announce = msg => Console.WriteLine($">>> {msg}");
announce("Смена окончена");
```
