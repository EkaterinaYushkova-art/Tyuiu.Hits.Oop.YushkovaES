### Система учёта животных в зоопарке (ZooCare)

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Финальная версия **ZooCare**: полиморфизм с перекрытием и перегрузкой методов, а также generic-коллекция `AnimalCollection<T>` для учёта животных в вольерах зоопарка (`ZooAnimal`, `Mammal`, `Bird`, `Reptile`, `Enclosure`, `Zoo`).

#### Дополнительное задание
Добавьте к существующим классам (базовому и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и перегрузкой методов, а также generic классы

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public ZooAnimal(string name, string species, int age, double weight)
    {
        Name = name;
        Species = species;
        Age = age;
        Weight = weight;
    }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт звук");
    }

    // перегрузка
    public void MakeSound(int times)
    {
        for (int i = 0; i < times; i++)
            MakeSound();
    }

    public void MakeSound(string mood)
    {
        Console.WriteLine($"{Name} издаёт звук в настроении «{mood}»");
    }

    public virtual string GetInfo() => $"{Species} {Name}, {Age} л., {Weight} кг";
}

public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, double weight, bool hasFur)
        : base(name, species, age, weight)
    {
        HasFur = hasFur;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: звук млекопитающего");
}

public class Bird : ZooAnimal
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double weight, double wingspan)
        : base(name, species, age, weight)
    {
        Wingspan = wingspan;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Чирик!");
}

public class Reptile : ZooAnimal
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, double weight, bool isVenomous)
        : base(name, species, age, weight)
    {
        IsVenomous = isVenomous;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Ш-ш-ш");
}

public class AnimalCollection<T> where T : ZooAnimal
{
    private List<T> _items = new List<T>();

    public void Add(T animal) => _items.Add(animal);

    public IEnumerable<T> Items => _items;

    public int Count => _items.Count;

    public void ForEachSound()
    {
        foreach (var a in _items)
            a.MakeSound();
    }
}

public class Enclosure
{
    public string Title { get; set; }
    public AnimalCollection<ZooAnimal> Animals { get; } = new AnimalCollection<ZooAnimal>();

    public Enclosure(string title) { Title = title; }

    public void Add(ZooAnimal animal)
    {
        Animals.Add(animal);
        Console.WriteLine($"[{Title}] + {animal.GetInfo()}");
    }
}

public class Zoo
{
    public string Name { get; set; }
    public List<Enclosure> Enclosures { get; } = new List<Enclosure>();

    public Zoo(string name) { Name = name; }
    public void AddEnclosure(Enclosure e) => Enclosures.Add(e);

    public void Show()
    {
        Console.WriteLine($"=== {Name} ===");
        foreach (var e in Enclosures)
        {
            Console.WriteLine($"Вольер {e.Title} ({e.Animals.Count}):");
            foreach (var a in e.Animals.Items)
            {
                Console.WriteLine("  " + a.GetInfo());
                a.MakeSound();
            }
        }
    }
}

Zoo zoo = new Zoo("ZooCare Ultimate");
var mammals = new Enclosure("Млекопитающие");
var birds = new Enclosure("Птицы");
var reptiles = new Enclosure("Рептилии");

var dog = new Mammal("Рекс", "Собака", 5, 30, true);
var cat = new Mammal("Мурка", "Кошка", 3, 4, true);
var parrot = new Bird("Кеша", "Попугай", 2, 0.4, 0.35);
var snake = new Reptile("Каа", "Питон", 8, 15, false);

mammals.Add(dog);
mammals.Add(cat);
birds.Add(parrot);
reptiles.Add(snake);

dog.MakeSound(2);
cat.MakeSound("спокойствие");
parrot.MakeSound();

zoo.AddEnclosure(mammals);
zoo.AddEnclosure(birds);
zoo.AddEnclosure(reptiles);
zoo.Show();

AnimalCollection<ZooAnimal> all = new AnimalCollection<ZooAnimal>();
all.Add(dog);
all.Add(cat);
all.Add(parrot);
all.Add(snake);
Console.WriteLine($"Generic-коллекция содержит: {all.Count}");
all.ForEachSound();
```
