#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы


<h2 style="color:DodgerBlue">Реализация:</h2>

```
public abstract class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }

    protected ZooAnimal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
    }

    public virtual void Speak()
    {
        Console.WriteLine($"{Name} издаёт звук");
    }

    // перегрузка
    public void Speak(int times)
    {
        for (int i = 0; i < times; i++)
            Speak();
    }

    public void Speak(string emotion)
    {
        Console.WriteLine($"{Name} выражает: {emotion}");
    }

    public virtual void Describe() =>
        Console.WriteLine($"{Name} ({Species}), {Age} лет");
}

public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, bool hasFur)
        : base(name, species, age) => HasFur = hasFur;

    public override void Speak() => Console.WriteLine($"Млекопитающее {Name}: рычит");
}

public class Bird : ZooAnimal
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double wingspan)
        : base(name, species, age) => Wingspan = wingspan;

    public override void Speak() => Console.WriteLine($"Птица {Name}: поёт");

    public override void Describe()
    {
        base.Describe();
        Console.WriteLine($"Размах крыльев: {Wingspan}");
    }
}

public class Reptile : ZooAnimal
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, bool isVenomous)
        : base(name, species, age) => IsVenomous = isVenomous;

    public override void Speak() => Console.WriteLine($"Рептилия {Name}: шипит");
}

public class ZooCollection<T> where T : ZooAnimal
{
    private readonly List<T> items = new();

    public void Add(T item) => items.Add(item);

    public void ForEachSpeak()
    {
        foreach (var item in items)
            item.Speak();
    }

    public IEnumerable<T> Items => items;
}

public class Enclosure
{
    public string Title { get; set; }
    public ZooCollection<ZooAnimal> Animals { get; } = new();

    public Enclosure(string title) => Title = title;
}

var enclosure = new Enclosure("ZooCare Main");
enclosure.Animals.Add(new Mammal("Симба", "Лев", 5, true));
enclosure.Animals.Add(new Bird("Киви", "Страус", 3, 2.1));
enclosure.Animals.Add(new Reptile("Гена", "Крокодил", 10, false));

Console.WriteLine($"Вольер: {enclosure.Title}");
foreach (var animal in enclosure.Animals.Items)
    animal.Describe();

enclosure.Animals.ForEachSpeak();

var animalsList = new List<ZooAnimal>(enclosure.Animals.Items);
var first = animalsList[0];
first.Speak(2);
first.Speak("голод");
```
