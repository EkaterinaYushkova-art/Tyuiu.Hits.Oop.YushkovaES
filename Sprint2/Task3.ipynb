#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование


<h2 style="color:DodgerBlue">Реализация:</h2>

```
// ZooCare: простое, многоуровневое и множественное наследование
public abstract class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }

    protected ZooAnimal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
    }

    public virtual void Describe() =>
        Console.WriteLine($"{Name} ({Species}), {Age} лет");
}

public interface IFeedable
{
    void Feed(string food);
}

public interface ITrainable
{
    void Train(string command);
}

// Простое наследование
public class Reptile : ZooAnimal, IFeedable
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, bool isVenomous)
        : base(name, species, age) => IsVenomous = isVenomous;

    public void Feed(string food) => Console.WriteLine($"{Name} ест {food}");

    public override void Describe()
    {
        base.Describe();
        Console.WriteLine($"Ядовитый: {IsVenomous}");
    }
}

// Многоуровневое: ZooAnimal -> Mammal -> Lion
public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, bool hasFur)
        : base(name, species, age) => HasFur = hasFur;

    public override void Describe()
    {
        base.Describe();
        Console.WriteLine($"Шерсть: {HasFur}");
    }
}

public class Lion : Mammal, IFeedable, ITrainable
{
    public int PrideSize { get; set; }

    public Lion(string name, int age, int prideSize)
        : base(name, "Лев", age, true) => PrideSize = prideSize;

    public void Feed(string food) => Console.WriteLine($"Лев {Name} ест {food}");
    public void Train(string command) => Console.WriteLine($"Лев {Name} выполняет: {command}");

    public override void Describe()
    {
        base.Describe();
        Console.WriteLine($"Размер прайда: {PrideSize}");
    }
}

// Множественное через интерфейсы
public class Bird : ZooAnimal, IFeedable, ITrainable
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double wingspan)
        : base(name, species, age) => Wingspan = wingspan;

    public void Feed(string food) => Console.WriteLine($"Птица {Name} клюёт {food}");
    public void Train(string command) => Console.WriteLine($"Птица {Name}: {command}");
}

public class Enclosure
{
    public string Title { get; set; }
    public List<ZooAnimal> Animals { get; } = new();

    public Enclosure(string title) => Title = title;
    public void Add(ZooAnimal animal) => Animals.Add(animal);
}

var enclosure = new Enclosure("Африка");
var lion = new Lion("Симба", 5, 8);
var bird = new Bird("Киви", "Страус", 4, 2.0);
var reptile = new Reptile("Гена", "Крокодил", 12, false);

enclosure.Add(lion);
enclosure.Add(bird);
enclosure.Add(reptile);

foreach (var animal in enclosure.Animals)
    animal.Describe();

lion.Feed("мясо");
lion.Train("сидеть");
bird.Feed("зерно");
reptile.Feed("рыба");
```
