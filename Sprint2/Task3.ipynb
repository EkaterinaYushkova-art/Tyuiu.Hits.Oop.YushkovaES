### Система учёта животных в зоопарке (ZooCare)

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

В **ZooCare** добавлены уровни наследования: `ZooAnimal` → `Mammal` → `Dog`/`Cat`, а также `Bird` и `Reptile`.  
Реализованы простое и сложное наследование и множественное наследование через интерфейсы (`IFeedable`, `ITrainable`).

#### Дополнительное задание
Добавьте к существующим классам (базовому и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public interface IFeedable
{
    void Feed(string food);
}

public interface ITrainable
{
    void Train(string command);
}

public class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }
    public double Weight { get; set; }

    public ZooAnimal(string name, string species, int age, double weight)
    {
        Name = name;
        Species = species;
        Age = age;
        Weight = weight;
    }

    public virtual void MakeSound() => Console.WriteLine($"{Name} издаёт звук");
    public virtual string GetInfo() => $"{Species} {Name}, {Age} л., {Weight} кг";
}

public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, double weight, bool hasFur)
        : base(name, species, age, weight)
    {
        HasFur = hasFur;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: звук млекопитающего");
}

public class Dog : Mammal, IFeedable, ITrainable
{
    public string Breed { get; set; }

    public Dog(string name, int age, double weight, string breed)
        : base(name, "Собака", age, weight, true)
    {
        Breed = breed;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Гав!");
    public void Feed(string food) => Console.WriteLine($"{Name} ест {food}");
    public void Train(string command) => Console.WriteLine($"{Name} выполняет команду «{command}»");
}

public class Cat : Mammal, IFeedable
{
    public string Color { get; set; }

    public Cat(string name, int age, double weight, string color)
        : base(name, "Кошка", age, weight, true)
    {
        Color = color;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Мяу!");
    public void Feed(string food) => Console.WriteLine($"{Name} ест {food}");
}

public class Bird : ZooAnimal, IFeedable
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double weight, double wingspan)
        : base(name, species, age, weight)
    {
        Wingspan = wingspan;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Чирик!");
    public void Feed(string food) => Console.WriteLine($"{Name} клюёт {food}");
}

public class Reptile : ZooAnimal
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, double weight, bool isVenomous)
        : base(name, species, age, weight)
    {
        IsVenomous = isVenomous;
    }

    public override void MakeSound() => Console.WriteLine($"{Name}: Ш-ш-ш");
}

public class Enclosure
{
    public string Title { get; set; }
    public List<ZooAnimal> Animals { get; } = new List<ZooAnimal>();

    public Enclosure(string title) { Title = title; }

    public void Add(ZooAnimal animal)
    {
        Animals.Add(animal);
        Console.WriteLine($"[{Title}] + {animal.GetInfo()}");
    }
}

public class Zoo
{
    public string Name { get; set; }
    public List<Enclosure> Enclosures { get; } = new List<Enclosure>();

    public Zoo(string name) { Name = name; }
    public void AddEnclosure(Enclosure e) => Enclosures.Add(e);

    public void Show()
    {
        Console.WriteLine($"Зоопарк {Name}");
        foreach (var e in Enclosures)
        {
            Console.WriteLine($" Вольер {e.Title}:");
            foreach (var a in e.Animals)
            {
                Console.WriteLine("  " + a.GetInfo());
                a.MakeSound();
            }
        }
    }
}

Zoo zoo = new Zoo("ZooCare");
var dogs = new Enclosure("Собаки");
var cats = new Enclosure("Кошки");
var birds = new Enclosure("Птицы");
var reptiles = new Enclosure("Рептилии");

var dog = new Dog("Рекс", 5, 28, "Овчарка");
var cat = new Cat("Мурка", 3, 4.5, "Рыжий");
var bird = new Bird("Кеша", "Попугай", 2, 0.4, 0.35);
var reptile = new Reptile("Гена", "Крокодил", 10, 120, false);

dogs.Add(dog);
cats.Add(cat);
birds.Add(bird);
reptiles.Add(reptile);

dog.Feed("корм");
dog.Train("сидеть");
cat.Feed("рыба");
bird.Feed("зерно");

zoo.AddEnclosure(dogs);
zoo.AddEnclosure(cats);
zoo.AddEnclosure(birds);
zoo.AddEnclosure(reptiles);
zoo.Show();
```
