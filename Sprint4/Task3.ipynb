#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) создайте явную реализации интерфейса и управление зависимостями 


<h2 style="color:DodgerBlue">Реализация:</h2>

```
public interface IFeedable
{
    void Feed(string food);
}

public interface ITrainable
{
    void Feed(string food);
}

public interface ICareService
{
    void Apply(string animalName);
}

public class VetService : ICareService
{
    public void Apply(string animalName) => Console.WriteLine($"Вет-сервис: осмотр {animalName}");
}

public class TrainerService : ICareService
{
    public void Apply(string animalName) => Console.WriteLine($"Тренер работает с {animalName}");
}

public abstract class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }
    protected readonly ICareService Care;

    protected ZooAnimal(string name, string species, int age, ICareService care)
    {
        Name = name;
        Species = species;
        Age = age;
        Care = care;
    }

    public virtual void Describe() => Console.WriteLine($"{Name} ({Species}), {Age} лет");
    public void ReceiveCare() => Care.Apply(Name);
}

public class Mammal : ZooAnimal, IFeedable, ITrainable
{
    public bool HasFur { get; set; }

    public Mammal(string name, string species, int age, bool hasFur, ICareService care)
        : base(name, species, age, care) => HasFur = hasFur;

    void IFeedable.Feed(string food) => Console.WriteLine($"IFeedable: {Name} ест {food}");
    void ITrainable.Feed(string food) => Console.WriteLine($"ITrainable: {Name} получает лакомство {food} за команду");

    public override void Describe()
    {
        base.Describe();
        Console.WriteLine($"Шерсть: {HasFur}");
    }
}

public class Bird : ZooAnimal, IFeedable, ITrainable
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double wingspan, ICareService care)
        : base(name, species, age, care) => Wingspan = wingspan;

    void IFeedable.Feed(string food) => Console.WriteLine($"IFeedable: птица {Name} клюёт {food}");
    void ITrainable.Feed(string food) => Console.WriteLine($"ITrainable: птица {Name} учится за {food}");
}

public class Reptile : ZooAnimal, IFeedable
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, bool isVenomous, ICareService care)
        : base(name, species, age, care) => IsVenomous = isVenomous;

    public void Feed(string food) => Console.WriteLine($"Рептилия {Name} ест {food}");
}

public class Enclosure
{
    public string Title { get; set; }
    public List<ZooAnimal> Animals { get; } = new();

    public Enclosure(string title) => Title = title;
    public void Add(ZooAnimal animal) => Animals.Add(animal);
}

var enclosure = new Enclosure("ZooCare — Африка");
var lion = new Mammal("Симба", "Лев", 5, true, new TrainerService());
var bird = new Bird("Киви", "Страус", 3, 2.0, new VetService());
var reptile = new Reptile("Гена", "Крокодил", 12, false, new VetService());

enclosure.Add(lion);
enclosure.Add(bird);
enclosure.Add(reptile);

foreach (var animal in enclosure.Animals)
{
    animal.Describe();
    animal.ReceiveCare();
}

((IFeedable)lion).Feed("мясо");
((ITrainable)lion).Feed("кусочек");
((IFeedable)bird).Feed("зерно");
reptile.Feed("рыба");
```
