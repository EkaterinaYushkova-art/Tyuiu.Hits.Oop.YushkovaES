<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

**Система учёта животных в зоопарке (ZooCare)** — учебный проект для учёта животных, вольеров и зоопарка.
Базовый класс `ZooAnimal` и производные `Mammal`, `Bird`, `Reptile`; классы `Enclosure` и `Zoo` связывают животных и места содержания.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
// ZooCare — базовая модель зоопарка
public class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }

    public ZooAnimal(string name, string species, int age)
    {
        Name = name;
        Species = species;
        Age = age;
    }

    public virtual void Describe()
    {
        Console.WriteLine($"{Name}, вид: {Species}, возраст: {Age}");
    }
}

public class Mammal : ZooAnimal
{
    public Mammal(string name, string species, int age) : base(name, species, age) { }
}

public class Bird : ZooAnimal
{
    public double Wingspan { get; set; }

    public Bird(string name, string species, int age, double wingspan)
        : base(name, species, age)
    {
        Wingspan = wingspan;
    }

    public override void Describe()
    {
        base.Describe();
        Console.WriteLine($"Размах крыльев: {Wingspan} м");
    }
}

public class Reptile : ZooAnimal
{
    public bool IsVenomous { get; set; }

    public Reptile(string name, string species, int age, bool isVenomous)
        : base(name, species, age)
    {
        IsVenomous = isVenomous;
    }
}

public class Enclosure
{
    public string Title { get; set; }
    public List<ZooAnimal> Animals { get; } = new();

    public Enclosure(string title) => Title = title;

    public void Add(ZooAnimal animal) => Animals.Add(animal);
}

public class Zoo
{
    public string Name { get; set; }
    public List<Enclosure> Enclosures { get; } = new();

    public Zoo(string name) => Name = name;

    public void AddEnclosure(Enclosure enclosure) => Enclosures.Add(enclosure);

    public void ShowAll()
    {
        Console.WriteLine($"Зоопарк: {Name}");
        foreach (var enclosure in Enclosures)
        {
            Console.WriteLine($"Вольер: {enclosure.Title}");
            foreach (var animal in enclosure.Animals)
                animal.Describe();
        }
    }
}

var zoo = new Zoo("ZooCare");
var savanna = new Enclosure("Саванна");
savanna.Add(new Mammal("Симба", "Лев", 5));
savanna.Add(new Bird("Киви", "Страус", 3, 2.1));
savanna.Add(new Reptile("Гена", "Крокодил", 12, false));
zoo.AddEnclosure(savanna);
zoo.ShowAll();
```
