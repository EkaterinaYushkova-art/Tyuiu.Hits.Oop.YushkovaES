### Система учёта животных в зоопарке (ZooCare)

<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Проект **ZooCare** — система учёта животных в зоопарке.  
Базовый класс `ZooAnimal` описывает общих животных зоопарка. От него наследуются `Mammal`, `Bird` и `Reptile`.  
Класс `Enclosure` хранит животных в вольере (композиция/агрегация), а класс `Zoo` объединяет вольеры и позволяет вести учёт поголовья.

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;
using System.Collections.Generic;

public class ZooAnimal
{
    public string Name { get; set; }
    public string Species { get; set; }
    public int Age { get; set; }

    public virtual void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт звук");
    }

    public virtual string GetInfo()
    {
        return $"{Species} {Name}, возраст {Age}";
    }
}

public class Mammal : ZooAnimal
{
    public bool HasFur { get; set; } = true;

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} (млекопитающее) рычит или мяукает");
    }
}

public class Bird : ZooAnimal
{
    public double Wingspan { get; set; }

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} (птица) чирикает");
    }
}

public class Reptile : ZooAnimal
{
    public bool IsColdBlooded { get; set; } = true;

    public override void MakeSound()
    {
        Console.WriteLine($"{Name} (рептилия) шипит");
    }
}

public class Enclosure
{
    public string Title { get; set; }
    public List<ZooAnimal> Animals { get; private set; } = new List<ZooAnimal>();

    public Enclosure(string title)
    {
        Title = title;
    }

    public void AddAnimal(ZooAnimal animal)
    {
        Animals.Add(animal);
        Console.WriteLine($"В вольер «{Title}» добавлен: {animal.GetInfo()}");
    }
}

public class Zoo
{
    public string Name { get; set; }
    public List<Enclosure> Enclosures { get; private set; } = new List<Enclosure>();

    public Zoo(string name)
    {
        Name = name;
    }

    public void AddEnclosure(Enclosure enclosure)
    {
        Enclosures.Add(enclosure);
    }

    public void ShowAll()
    {
        Console.WriteLine($"Зоопарк: {Name}");
        foreach (var enclosure in Enclosures)
        {
            Console.WriteLine($" Вольер: {enclosure.Title}");
            foreach (var animal in enclosure.Animals)
            {
                Console.WriteLine("  - " + animal.GetInfo());
                animal.MakeSound();
            }
        }
    }
}

Zoo zoo = new Zoo("Уральский зоопарк");
Enclosure mammals = new Enclosure("Млекопитающие");
Enclosure birds = new Enclosure("Птицы");
Enclosure reptiles = new Enclosure("Рептилии");

mammals.AddAnimal(new Mammal { Name = "Мурка", Species = "Кошка", Age = 4, HasFur = true });
mammals.AddAnimal(new Mammal { Name = "Рекс", Species = "Собака", Age = 5, HasFur = true });
birds.AddAnimal(new Bird { Name = "Кеша", Species = "Попугай", Age = 2, Wingspan = 0.4 });
reptiles.AddAnimal(new Reptile { Name = "Гена", Species = "Крокодил", Age = 12 });

zoo.AddEnclosure(mammals);
zoo.AddEnclosure(birds);
zoo.AddEnclosure(reptiles);
zoo.ShowAll();
```
